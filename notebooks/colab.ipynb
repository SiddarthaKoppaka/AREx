```python
if DATASET_ROOT is None:
    from arc_agi import Arcade, OperationMode

    DATASET_ROOT = Path("/content/arc_data/environment_files")
    DATASET_ROOT.mkdir(parents=True, exist_ok=True)

    print("Downloading the official ls20 environment...")
    bootstrap = Arcade(
        operation_mode=OperationMode.NORMAL,
        environments_dir=str(DATASET_ROOT),
    )
    downloaded = bootstrap.make(
        "ls20",
        seed=0,
        include_frame_data=True,
    )

    if downloaded is None:
        raise RuntimeError(
            "ARC download failed. Configure ARC_API_KEY and rerun this cell."
        )

print("ARC environment root:", DATASET_ROOT)
```
